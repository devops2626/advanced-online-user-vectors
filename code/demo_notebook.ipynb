# Interactive Demo: Retrieval + MMR

This notebook lets you explore the toy recommendation pipeline interactively.

**What you can try:**
- Change the number of candidates and MMR λ
- Apply metadata filters (category / language)
- Observe how diversity changes the feed

```python
import numpy as np
from toy_data import generate_toy_catalog, generate_user_vector, filter_catalog
from retrieve_and_mmr import retrieve_candidates, maximal_marginal_relevance

# Generate data
catalog = generate_toy_catalog()
user_vector = generate_user_vector()
print(f"Catalog size: {len(catalog)} videos")
print(f"Embedding dim: {len(user_vector)}")
```

## 1. Pure Relevance Retrieval

```python
candidates = retrieve_candidates(user_vector, catalog, top_k=12)
print("Top candidates by cosine similarity:")
for i, item in enumerate(candidates, 1):
    print(f"{i:2d}. {item['video_id']} | {item['creator_id']} | {item['category']} | {item['language']} | score={item['ml_score']:.3f}")
```

## 2. Retrieval with Metadata Filters

```python
filtered = retrieve_candidates(
    user_vector, catalog, top_k=10, category="gaming", language="en"
)
print("Filtered results (gaming + en):")
for i, item in enumerate(filtered, 1):
    print(f"{i:2d}. {item['video_id']} | {item['creator_id']} | {item['category']} | {item['language']} | score={item['ml_score']:.3f}")
```

## 3. Apply MMR Diversity

```python
LAMBDA = 0.7  # higher = more relevance, lower = more diversity

diversified = maximal_marginal_relevance(candidates, top_n=8, lambda_param=LAMBDA)
print(f"Diversified feed (λ={LAMBDA}):")
for i, item in enumerate(diversified, 1):
    print(f"{i:2d}. {item['video_id']} | {item['creator_id']} | {item['category']} | score={item['ml_score']:.3f}")

print("\nCreator distribution:")
print("  Before:", [c['creator_id'] for c in candidates[:8]])
print("  After: ", [c['creator_id'] for c in diversified])
```

## 4. Experiment with λ

Try different values of `LAMBDA` (0.0 – 1.0) in the cell above and re-run to see the relevance–diversity trade-off.
